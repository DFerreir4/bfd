```python
class Singleton:
    _instance = None  # Variável de classe para armazenar a instância

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:  # Verifica se a instância já existe
            # Se não existe, cria uma nova usando o método __new__ da superclasse
            cls._instance = super(Singleton, cls).__new__(cls, *args, **kwargs)
        return cls._instance  # Retorna a instância única

# Uso da classe
s1 = Singleton()
s2 = Singleton()

print(f's1 é a mesma instância que s2? {s1 is s2}')

# Demonstrando que os atributos são compartilhados
s1.data = "Dados compartilhados"
print(f's2.data: {s2.data}')
```
