```python
#4

def calcular_imc(peso,altura):
    
    imc = peso/(altura**2)

    if imc < 18.5:
        return f"Abaixo do Peso! Seu IMC foi : {imc:,.2f}"
    elif imc >= 18.5 or imc < 24.9:
        return f"Peso Normal! Seu IMC foi : {imc:,.2f}"
    elif imc == 25 or imc <= 29.9:
        return f"Sobrepeso! Seu IMC foi : {imc:,.2f}"
    elif imc >= 30:
        return f"Obesidade! Seu IMC foi : {imc:,.2f}"
    

calcular_imc(70,1.75)
```

```python
#5

class Conta:

    def __init__(self,saldo=0):
        self.saldo = saldo

    def depositar(self,valor):
        self.saldo += valor
        return f"Foi depositado o valor de R$ {valor:.2f}"

    def sacar(self,valor):
        if valor > 0:
            if valor <= self.saldo:
                self.saldo -= valor
                return f"Você sacou R$ {valor:.2f}. Seu saldo é R$ {self.saldo}"
            else:
                return f"Sua conta está com o saldo R$ {self.saldo:.2f}. Não é possivel sacar o valor R$ {valor:.2f}."

c = Conta()
print("-"*30)
print(f"Saldo Inicial: {c.saldo}")
print("-"*30)
c.depositar(100)
print(f"+ Credito : R$ 100,00")
c.sacar(30)
print(f"- Debito : R$ 30,00")
print("-"*30)
print(f"Saldo Total: R$ {c.saldo:,.2f}")
print("-"*30)


```

```python
#6

class Pedido:
    
    def __init__(self,produto_p,quantidade_p):
        self.produto_p = produto_p
        self.quantidade_p = quantidade_p

class Cliente:

    def __init__(self,nome):
        self.nome = nome
    
    def fazer_pedido(self,produto,quantidade):
        if (produto == "" or produto == 0) or (quantidade == "" or quantidade == 0):
            print("Produto ou quantidade com valor incorreto")
        else: print(f"Cliente : {self.nome}")
        print(f"Produto : {produto} // Quantidade : {quantidade}")
        
      
print("-"*30)
c1 = Cliente("João")
p1 = Pedido("Notebook",2)
c1.fazer_pedido(p1.produto_p,p1.quantidade_p)
print("-"*30)
print("-"*30)
c2 = Cliente("Paulo")
p2 = Pedido("Celular",10)
c2.fazer_pedido(p2.produto_p,p2.quantidade_p)

```

```python
#7

class Ponto:

    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"{self.x} e {self.y}"
    
    def __add__(self):
        return self.x+self.y
    
p = Ponto(12,9)
p.__str__()
p.__add__()
```
