```python
#Controle de presença dos alunos
"""
Foi criado uma lista dos meses em que as escolas recebem os alunos para as aulas. Notem que está faltando os meses [Julho , Janeiro e Dezembro]. Esses meses são férias escolar. """
meses = [
    "[FEVEREIRO]", "[MARÇO]", "[ABRIL]", "[MAIO]", "[JUNHO]", "[AGOSTO]","[SETEMBRO]", "[OUTUBRO]", "[NOVEMBRO]"
]

"""
Os meses foram inseridos no while para que o usuário insira os dados corretamente nos seus respectivos meses.
"""
total_dias = 0
contador = 0
while contador < len(meses):
    qtd_dias = int(input(f"Digite a quantidade de presenças do mês de {meses[contador]} : "))
    total_dias += qtd_dias
    contador += 1

"""
Finalizando o laço. Os dados coletados vão passar pelo calculo de frequencia e também por porcentagem. Calculando o total de dias dividindo por 75%. A porcentagem é 
"""
frequencia = total_dias*0.75
porcentagem = (frequencia/200)*100

if frequencia >= 150:
    print("[APROVADO]")
    print("Aluno aprovado por ter comprido presença em sala de aula no ano letivo")
    print(f"Quantidade de dias: {int(frequencia)} // Porcentagem de presença: {int(porcentagem)}%.")
else:
    print("[REPROVADO]")
    print("Aluno reprovado por falta de presença em sala de aula no ano letivo.")
    print(f"Quantidade de dias: {int(frequencia)} // Porcentagem de presença: {int(porcentagem)}%.")

"""
Conforme a Lei de Diretrizes e Bases da Educação Nacional (LDB), a porcentagem mínima de presença de alunos em sala de aula para aprovação é de 75%. Isso significa que, em um ano letivo com 200 dias, um aluno não pode ter mais que 25% de faltas (50 dias letivos no total). 
"""

    
```
